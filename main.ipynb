```python
x_test=[[38.0,"female","High School",123335.0,16,"MORTGAGE","EDUCATION",14.06,11.0,640,"No"]]
x_test=pd.DataFrame(x_test,columns=['person_age','person_gender','person_education', 'person_income',
       'person_emp_exp', 'person_home_ownership', 'loan_intent',
       'loan_int_rate','cb_person_cred_hist_length',
       'credit_score', 'previous_loan_defaults_on_file'])

education_map={'Associate':4, 'Bachelor':1, 'Doctorate':3, 'High School':0, 'Master':2}
x_test['person_education']=x_test['person_education'].map(education_map)

previous_loan_defaults_on_file_map={'No':0,'Yes':1}
x_test['previous_loan_defaults_on_file']=x_test['previous_loan_defaults_on_file'].map(previous_loan_defaults_on_file_map)

#-------------------------
# مشخص کردن همه حالت‌های ممکن
# -------------------------

x_test["person_gender"] = pd.Categorical(
    x_test["person_gender"],
    categories=["female", "male"]
)

x_test["person_home_ownership"] = pd.Categorical(
    x_test["person_home_ownership"],
    categories=["MORTGAGE", "OTHER", "OWN", "RENT"]
)

x_test["loan_intent"] = pd.Categorical(
    x_test["loan_intent"],
    categories=[
        "DEBTCONSOLIDATION",
        "EDUCATION",
        "HOMEIMPROVEMENT",
        "MEDICAL",
        "PERSONAL",
        "VENTURE"
    ]
)


# -------------------------
# One Hot Encoding
# -------------------------

x_test = pd.get_dummies(
    x_test,
    columns=[
        "person_gender",
        "person_home_ownership",
        "loan_intent"
    ],
    dtype=int
)

print(x_test)
```
